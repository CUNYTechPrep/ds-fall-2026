## Exercise 2 - Filtering

Return the following subsets of the dataframe.

1. How many listings are there with a price less than 100? 


2. Find how many listings there are in just Brooklyn.


3. Find how many listings there are in Brooklyn with a price less than 100.


4. Using `.isin()` select anyone that has the host name of Michael, David, John, and Daniel.


5. Create a new column called `adjusted_price` that has $100 added to every listing in Williamsburg.  The prices for all other listings should be the same as the were before. 


6. What % of the rooms are private, and what % of the rooms are shared.  
    * Hint, use `.value_counts()`


```python
# 1. How many listings are there with a price less than 100? 
df = pd.read_csv("data/listings.csv")
df[df["price"] < 100].shape[0]

```

```python
# 2. Make a new DataFrame of listings in Brooklyn named `df_bk` 
# and find how many listings in just Brooklyn.

df_bk = df[df["neighbourhood_group"] == "Brooklyn"]
df_bk.shape[0]
```

```python
# 3. Find how many listings there are in Brooklyn with a price less than 100.
df[(df["neighbourhood_group"] == "Brooklyn") & (df["price"] < 100)].shape[0]


```

```python
# 4. Using `.isin()` select anyone that has the host name of Michael, David, John, and Daniel.
# How many total are there that have those names
df[df["host_name"].isin(["Michael","David","John","Daniel"])].shape[0]
```

```python
# 5. Create a new column called `adjusted_price` that has $100 added to every listing in Williamsburg.  
# The prices for all other listings should be the same as the were before. 
prior = df.shape[0]
df["adjusted_price"] = df["price"] + (df["neighbourhood"] == "Williamsburg") * 100
williamsburg_count = df[df["neighbourhood"] == "Williamsburg"][["neighbourhood", "price", "adjusted_price"]].shape[0]
other_count = df[df["neighbourhood"] != "Williamsburg"][["neighbourhood", "price", "adjusted_price"]].shape[0]
print("Total count:", prior)
print("Williamsburg count:", williamsburg_count)
print("Other count:", other_count)
print("Sum of Williamsburg and Other counts:", williamsburg_count + other_count)
print("Counts match total:", prior == williamsburg_count + other_count)


```

```python
# 6. What % of the rooms are private, and what % of the rooms are shared.  
percentages = df["room_type"].value_counts(normalize=True)
private = percentages.get("Private room", 0)
shared = percentages.get("Shared room", 0)
print("Private room %:", private * 100)
print("Shared room %:", shared * 100)
```

### 3. (Medium) Use the grammys.csv data for the next section of questions.

1. Who was won Album of the Year in 2016?


2. Who won Best Rap Album in 2009?


3. How many awards was Kendrick Lamar nomiated for, and how many did he win...?

```python
grammys = pd.read_csv("data/grammys.csv")
# 1A
album_of_the_year_2016_winner = grammys.loc[
    (grammys["category"].eq("Album of the Year")) &
    (grammys["year"].eq(2016)) &
    (grammys["winner"].eq(True))
].copy()
album_of_the_year_2016_winner_worker = album_of_the_year_2016_winner["workers"].values[0]
print("The winner of Album of the Year in 2016 was: {0}".format(album_of_the_year_2016_winner_worker))

# 2A
best_rap_album_2009_winner = grammys.loc[
    (grammys["category"] == "Best Rap Album") &
    (grammys["year"] == 2009) &
    (grammys["winner"] == True)
].copy()
best_rap_album_2009_winner_worker = best_rap_album_2009_winner["workers"].values[0]
print("The winner of Best Rap Album in 2009 was: {0}".format(best_rap_album_2009_winner_worker))

# 3A
kendrick = grammys[grammys["nominee"].str.contains("Kendrick Lamar", na=False)]
kendrick_nominations = kendrick.shape[0]
kendrick_wins = kendrick[kendrick["winner"] == True].shape[0]

print("Kendrick Lamar nominations: {0}".format(kendrick_nominations))
print("Kendrick Lamar wins: {0}".format(kendrick_wins))
```

### (Hard) Load the Game Logs for 2022 and add the column names using a dictionary.  
* [Link to the data page](https://www.retrosheet.org/gamelogs/)
* [Link to the column names](https://procatinator.com/)
* haha, gotta find them yourself!

```python
import re
from pathlib import Path

fields_path = Path("data/glfields.txt")
game_log_path = Path("data/gl2022.txt")
if not fields_path.exists() or not game_log_path.exists():
    raise FileNotFoundError(
        "Place glfields.txt and gl2022.txt in the data folder before running this cell."
    )

content = fields_path.read_text(encoding="utf-8")
lines = content.splitlines()
field_names = {}
current_field = None
current_description = []

for line in lines:
    if not line.strip() or line.strip().startswith("Field"):
        continue

    match = re.match(r"\s*(\d+(?:-\d+)?)\s+(.*)", line)
    if match:
        if current_field is not None:
            field_description = " ".join(current_description).strip()
            if "-" in current_field:
                start, end = map(int, current_field.split("-", 1))
                short_name = re.split(r"\s+and\s+", field_description, maxsplit=1)[0][:30]
                for col_idx in range(start - 1, end):
                    field_names[col_idx] = short_name
            else:
                col_idx = int(current_field) - 1
                short_name = field_description.split(".")[0][:40].strip()
                field_names[col_idx] = short_name

        current_field = match.group(1)
        current_description = [match.group(2).strip()]
    elif current_field is not None and line.strip() and not line.strip().startswith("Field"):
        current_description.append(line.strip())

if current_field is not None:
    field_description = " ".join(current_description).strip()
    if "-" in current_field:
        start, end = map(int, current_field.split("-", 1))
        short_name = re.split(r"\s+and\s+", field_description, maxsplit=1)[0][:30]
        for col_idx in range(start - 1, end):
            field_names[col_idx] = short_name
    else:
        col_idx = int(current_field) - 1
        short_name = field_description.split(".")[0][:40].strip()
        field_names[col_idx] = short_name

raw_df = pd.read_table(game_log_path, sep=",", header=None)
rename_dict = {
    col_idx: field_names.get(col_idx, f"field_{col_idx + 1}")
    for col_idx in range(raw_df.shape[1])
}
df = raw_df.rename(columns=rename_dict)

print(f"Total records: {df.shape[0]}")
print(f"Total columns: {df.shape[1]}")
print("\nFirst 15 column names:")
for i, col in enumerate(df.columns[:15], start=1):
    print(f"  {i}: {col}")
```

### (Extra Hard) Download the files for the past 5 years into a new folder and add them all into one data frame using pandas, then save that new file.
* Try to not hard code in the file names. We want to do this programmatically because what if we want to add new/more file names in the future.

```python
import glob

files = sorted(glob.glob("data/gl20*.txt"))
frames = [pd.read_table(f, sep=",", header=None) for f in files]

combined_df = pd.concat(frames, ignore_index=True)
save_path = "data/combined.txt"
combined_df.to_csv(save_path, index=False)

print(f"Combined text file saved to {save_path}")
```
