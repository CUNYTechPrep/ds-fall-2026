# Exercise 3 - Grouping

1. Using `groupby`, count how many listings are in each neighbourhood_group.


2. Using `groupby`, find the mean price for each of the neighbourhood_groups. 


3. Using `groupby` and `.agg()`, find the min and max price for each of the neighbourhood_groups. 


4. Using `groupby`, find the median price for each room type in each neighbourhood_group.


5. Using `groupby` and `.agg()`, find the count, min, max, mean, median, and std of the prices for each room type in each neighbourhood_group.

```python
# 1. Using `groupby`, count how many listings are in each neighbourhood_group.

#Use count
gb = df.groupby('neighbourhood_group')
#gb['neighbourhood_group'].count().sum() #finds the total num of listings
gb['neighbourhood_group'].count() #counts num of listings in each borough, use neighbourhood_group col
```

```python
# 2. Using `groupby`, find the mean price for each of the neighbourhood_groups. 

#Use the prev cell gb for neighbourhood_group
gb['price'].mean()  #use the PRICE col
```

```python
# 2.5. Using `groupby`, find the mean price for each room_type. 

#create a new groupby, group by room_type
gb2 = df.groupby('room_type') 
gb2['price'].mean() #use the PRICE col
```

```python
# 3. Using `groupby` and `.agg()`, find the min and max price for each of the neighbourhood_groups. 

gb['price'].agg(['min', 'max']) #agg() takes a list of functions
```

```python
# 4. Using `groupby`, find the mean price for each room_type in each neighbourhood_group.

group_by_list = ['neighbourhood_group', 'room_type'] #put relevant col in list
gb3 = df.groupby(group_by_list) #group by those col (in the list)

gb3['price'].mean() #use the PRICE col
```

```python
# 5. Using `groupby` and `.agg()`, find the count, min, max, mean, median, and std of the prices 
# for each room type in each neighbourhood_group.


gb3['price'].agg(['count', 'min', 'max', 'mean', 'median', 'std']) #put the func in a list
```

```python
# 1. Load the `prices.csv` and the `n_listings.csv`

#need to go one folder up
df2 = pd.read_csv('../data/prices.csv') #load the prices.csv, store in df2 variable

df2
```

```python
#need to go one folder up
df3 = pd.read_csv('../data/n_listings.csv', sep = ';') #load the n_listings.csv, store in df3 variable

df3
```

```python
# 2. Do join that keeps all the records for each table.

#Outer Join to keep all records from both tables, use df2 and df3 from prev cells
all_records = pd.merge(df2,                      # Outer join
                      df3,
                      on = "neighbourhood_group",
                      how = "outer")
all_records
```

```python
# 3. Save your joined csv as `joined.csv` into the data folder. 

#need to go one folder up
new_file = '../data/joined.csv' #want to save in data folder
all_records.to_csv(new_file) #use name_of_df.to_csv(folder_name), used all_records bc that is the name of the dataframe

```

```python
# 4. Load your newly saved file, see if it looks the same.  If not, try saving with argument `index=False`

#use new_file --> name of the df file var
with open(new_file, 'rt') as text_file:
    for line in text_file.readlines():
        print(line)
```
