```python
# 1. How many listings are there with a price less than 100?

#df = pd.read_csv("../data/listings.csv")

#filter of true and false values for the price column
underHundred = df['price'] < 100 

#Only true values in new df
listingsUnderHundred = df[underHundred]

# .shape[0] will give you the number of rows in a dataframe
countUnderHundred = listingsUnderHundred.shape[0]
print(countUnderHundred)
df.columns
```

```python
# 2. Make a new DataFrame of listings in Brooklyn named `df_bk` 
# and find how many listings in just Brooklyn.

#df = pd.read_csv("../data/listings.csv")


df_bk = df[df['neighbourhood_group'] == 'Brooklyn']

#Me trying to make sense of this: 
#the dataframe specifically the neighborhood group that is equal to 'Brooklyn'
# and make a dataframe with just those rows

count_bk = df_bk.shape[0]
print(count_bk)
```

```python
# 3. Find how many listings there are in Brooklyn with a price less than 100.
bkPrices = df_bk['price']
listingsUnderHundred = df_bk[bkPrices < 100 ].shape[0]
print(listingsUnderHundred)
```

```python
# 4. Using `.isin()` select anyone that has the host name of Michael, David, John, and Daniel.
# How many total are there that have those names
df.head()
#df['host_name']
names = ['Michael', 'David', 'John', 'Daniel']
nameFilter= df['host_name'].isin(names)
filteredNames = df[nameFilter]
total = nameFilter.sum()
print(total)
```

```python
# 5. Create a new column called `adjusted_price` that has $100 added to every listing in Williamsburg.  
# The prices for all other listings should be the same as the were before. 

df['adjusted_price'] = df['price']
df['adjusted_price'] = np.where(df['neighbourhood'] == 'Williamsburg', df['price'] + 100 , df['price'])

df.head()
```

```python
# 6. What % of the rooms 
# are private, and what % of the rooms are shared.  

#in my search including looking at the .csv , I found that 
# 'Private room' is what I am looking for in 'room_type'
# also 'Shared room'

#df.columns
# NOTE: could have used the lines below in the lecture but this wouldn't isolate
# the two types of rooms.
# dfp = df.neighbourhood_group.value_counts(normalize=True) 
# dfp

type = df['room_type']
private = (type == 'Private room').sum()
shared = (type == 'Shared room').sum()

total = len(df)

privatePercent = private / total * 100
sharedPercent = shared / total * 100 

print( f"Private rooms make up {privatePercent} %  and shared rooms make up {sharedPercent} % " )

#print(shared
```

# Exercise 3 - Grouping

1. Using `groupby`, count how many listings are in each neighbourhood_group.


2. Using `groupby`, find the mean price for each of the neighbourhood_groups. 


3. Using `groupby` and `.agg()`, find the min and max price for each of the neighbourhood_groups. 


4. Using `groupby`, find the median price for each room type in each neighbourhood_group.


5. Using `groupby` and `.agg()`, find the count, min, max, mean, median, and std of the prices for each room type in each neighbourhood_group.

```python
# 1. Using `groupby`, count how many listings are in each neighbourhood_group.

df.groupby('neighbourhood_group').size()

```

```python
# 2. Using `groupby`, find the mean price for each of the neighbourhood_groups. 

df.groupby('neighbourhood_group')['price'].mean()
```

```python
# 2.5. Using `groupby`, find the mean price for each room_type. 
df.groupby('room_type')['price'].mean()
```

```python
# 3. Using `groupby` and `.agg()`, find the min and max price for each of the neighbourhood_groups. 

nhGroups = df.groupby('neighbourhood_group')
nhGroups['price'].agg(['min' , 'max'])
```

```python
# 4. Using `groupby`, find the mean price for each room_type in each neighbourhood_group.

nhGroups ['price'].agg(['mean'])
```

```python
# 5. Using `groupby` and `.agg()`, find the count, min, max, mean, median, and std of the prices 
# for each room type in each neighbourhood_group.

nhGroups = df.groupby('neighbourhood_group')
nhGroups['price'].agg(['count','min' , 'max', 'mean', 'median', 'std'])

```
