### Clean the Data

Well, we can’t get our answers with what we currently have, so you’ll have to make some changes.

Let’s not worry about anything below the first 5 values for now. Convert the top 5 to a single canonical value―say, `'US'`, which is nice and short.

```python
# Replace them all in-place with 'US'.
df_salary_tech["country"] = (
    df_salary_tech["country"]
    .astype(str)
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.replace(
        r"^(?:united states(?: of america)?|u\.?s\.?|usa|us|america|unite[ds]? states?)$",
        "US",
        regex=True,
        case=False
    )
)
df_salary_tech["country"].value_counts().head(15)
```

### Bonus #0

Apply the same steps as before to `df_salary`, but do not filter for any specific industry. Do perform the other data cleaning stuff, and get to a point where you can generate the minimum, mean, and maximum by state.

```python
# Bonus #0: same steps as before, but on df_salary without filtering for one industry.
df_salary_clean = df_salary.copy()
df_salary_clean = df_salary_clean[df_salary_clean["currency"] == "USD"].copy()
df_salary_clean["country"] = df_salary_clean["country"].astype(str).str.strip()
df_salary_clean["country"] = df_salary_clean["country"].str.replace(
    r".*(?:united states|u\.?s\.?|usa|america).*",
    "US",
    regex=True,
    case=False
)
df_salary_clean["salary"] = pd.to_numeric(df_salary_clean["salary"], errors="coerce")
df_salary_clean = df_salary_clean[(df_salary_clean["country"] == "US") & df_salary_clean["state"].notna()].copy()
salary_by_state_all = (
    df_salary_clean.dropna(subset=["state", "salary"])
    .groupby("state")["salary"]
    .agg(["min", "mean", "max"])
    .sort_values("mean", ascending=False)
)
salary_by_state_all.head(10)
```

```python
# Bonus #2: keep only rows whose state is a single state name, not a combined value like "California, Colorado."
df_single_state = (
    df_salary_clean
    .dropna(subset=["state"])
    .loc[~df_salary_clean["state"].str.contains(r"[,/] | and | & ", case=False, na=False)]
    .copy()
)
df_single_state[["state"]].head()
```

### Bonus #2

Filter out the non-single-states (e.g., `"California, Colorado"`) in the most elegant way possible (i.e., not by blacklisting all the bad values).


```python
# Bonus #3: show quantiles and counts by state.
state_quantiles = (
    df_salary_clean.dropna(subset=["state", "salary"])
    .groupby("state")["salary"]
    .quantile([0, 0.05, 0.25, 0.5, 0.75, 0.95, 1.0])
    .unstack()
    .sort_values(0.5, ascending=False)
)
state_quantiles["count"] = (
    df_salary_clean.dropna(subset=["state", "salary"])
    .groupby("state")["salary"]
    .count()
)
state_quantiles.head(10)
```
