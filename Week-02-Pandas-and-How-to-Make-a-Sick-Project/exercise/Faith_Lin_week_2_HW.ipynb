It’s looking good so far. Let’s find out the minimum, mean, and maximum (in that order) salary by state, sorted by the mean in descending order.

```python
# Find the minimum, mean, and maximum salary in USD by U.S. state.
```

```python
df_salary2.info() #salary values are of str datatype
```

Well, pooh! We forgot that `salary` isn’t numeric. Something wrong must be fixed.

```python
# Fix it.

df_salary2['salary'] = pd.to_numeric(df_salary2['salary'].str.replace(",", "")) #remove the commas in the num
df_salary2
```

Let’s try that again.

```python
# Try it again. Yeah!

gb2 = df_salary2.groupby('state') #group by the state, org df has states for US only
gb2['salary'].agg( ['min', 'mean', 'max'] ).sort_values(by='mean', ascending = False) #sort by mean ins DESC order
```

That did the trick! Now let’s narrow this to data 2021 and 2022 just because (lel). *(Hint: that timestamp column may not be a temporal type right now.)*

```python
# Filter the data to within 2021, 2022, or 2023, saving the DataFrame to a new variable, and generate the summary again.

df_salary2['timestamp'] = pd.to_datetime(df_salary2['timestamp']) #convert timestamp col to datetime datatype

df_salary2
```

```python
#create a NEW col to show the year only
df_salary2['year'] = df_salary2['timestamp'].dt.year #extract year only from datetime type

df_salary2
```

```python
df_salary2.info() #year is int datatype
```

```python
#create 2 conditions 
year1 = df_salary2['year'] == 2021
year2 = df_salary2['year'] == 2022

#create a new dataframe to filter for 2021 and 2022
df_salary_year = df_salary2[year1 | year2].copy() #use or operator because want data from 2021 or 2022, can't be both years at once
df_salary_year
```
