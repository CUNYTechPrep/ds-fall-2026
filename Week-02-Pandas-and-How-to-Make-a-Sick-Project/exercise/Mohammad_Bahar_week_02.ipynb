### Clean the Data

Perhaps we’d like to know more another day, but the team is really interested in just the relationship between calories (`calories_day`) and weight. …and maybe gender.

Can you remove the other columns? (Assign the result to a new variable, `df_food_col_subset`.)

```python
# 🚀 Extra credit: What about a third way to remove columns?

df_food_col_subset = df_food.loc[:, ['calories_day', 'weight', 'Gender']]
```

What about `NaN`s? How many are there?

We gotta remove those `NaN`s―the entire row.

```python
# Drop ‘em in-place.

df_food_col_subset.dropna(inplace=True)
```

```python
# 🚀 Extra credit: Check if the in-place modifications also affected the original DataFrame, df_food. 🙂

df_food.isna().sum()

# df_food still has its original NaNs -- df_food_col_subset was created with a column
# selection (a copy), so dropping rows on it in-place did not touch df_food.
```

But what about the weird non-numeric values in the column obviously meant for numeric data?

Notice the data type of that column from when you got the types of all the columns?

If only we could convert the column to a numeric type and drop the rows with invalid values. 🤔

```python
# Fix that in-place.

df_food_col_subset['weight'] = pd.to_numeric(df_food_col_subset['weight'], errors='coerce')
df_food_col_subset.dropna(subset=['weight'], inplace=True)
```

### Clean the Data

Well, we can’t get our answers with what we currently have, so you’ll have to make some changes.

Let’s not worry about anything below the first 5 values for now. Convert the top 5 to a single canonical value―say, `'US'`, which is nice and short.

```python
# Replace them all in-place with 'US'.

top5_us_values = ['United States', 'USA', 'US', 'U.S.', 'United States of America']
df_salary_tech['country'] = df_salary_tech['country'].replace(top5_us_values, 'US')
```

Have a look at the count of each unique country again now.

```python
# Count again.

df_salary_tech['country'].value_counts()
```

Did you notice anything interesting?

```python
# 🚀 Extra credit: Resolve [most of] those anomalous cases too without exhaustively taking every variant literally into account.

# Catch stray whitespace/case/typo variants of "US"/"USA"/"United States (of America)"
# with one regex instead of blacklisting every misspelling individually.
us_pattern = r'^u\.?s\.?a?\.?$|^unite[d]?\s*states?(\s*of\s*america)?$'
looks_like_us = (
    df_salary_tech['country']
    .astype('string')
    .str.strip()
    .str.lower()
    .str.match(us_pattern, na=False)
)
df_salary_tech.loc[looks_like_us, 'country'] = 'US'
```

```python

# 🚀 Extra credit: If you’ve resolved it, let’s see how well you did by counting the number of instances of each unique value.

df_salary_tech['country'].value_counts()
```

It’s looking good so far. Let’s find out the minimum, mean, and maximum (in that order) salary by state, sorted by the mean in descending order.

```python
# Find the minimum, mean, and maximum salary in USD by U.S. state.

df_salary_tech.groupby('state')['salary'].agg(['min', 'mean', 'max']).sort_values('mean', ascending=False)
```

 Well, pooh! We forgot that `salary` isn’t numeric. Something wrong must be fixed. 🤔

```python
# Fix it in-place.

df_salary_tech['salary'] = pd.to_numeric(
    df_salary_tech['salary'].astype(str).str.replace(',', '', regex=False),
    errors='coerce',
)
```

Let’s try that again.

```python
# Try it again. Yeah!

df_salary_tech.groupby('state')['salary'].agg(['min', 'mean', 'max']).sort_values('mean', ascending=False)
```

That did the trick! Now let’s narrow this to data 2021 and 2022 just because (lel). *(Hint: that timestamp column may not be a temporal type right now.)*

```python
# Filter the data to within 2021, 2022, or 2023, saving the DataFrame to a new variable, and generate the summary again.

df_salary_tech['timestamp'] = pd.to_datetime(df_salary_tech['timestamp'], errors='coerce')

df_salary_tech_recent = df_salary_tech[df_salary_tech['timestamp'].dt.year.isin([2021, 2022, 2023])]

df_salary_tech_recent.groupby('state')['salary'].agg(['min', 'mean', 'max']).sort_values('mean', ascending=False)
```

## Bonus

Clearly, we do not have enough data to produce useful figures for the level of specificity you’ve now reached. What do you notice about Delaware and West Virginia?

Let’s back out a bit and return to `df_salary` (which was the loaded data with renamed columns but *sans* filtering).

### Bonus #0

Apply the same steps as before to `df_salary`, but do not filter for any specific industry. Do perform the other data cleaning stuff, and get to a point where you can generate the minimum, mean, and maximum by state.

```python
# Bonus #0: Apply the same cleaning steps to df_salary, but without filtering by industry.

df_salary_all = df_salary[df_salary['currency'] == 'USD'].copy()

top5_us_values = ['United States', 'USA', 'US', 'U.S.', 'United States of America']
df_salary_all['country'] = df_salary_all['country'].replace(top5_us_values, 'US')

us_pattern = r'^u\.?s\.?a?\.?$|^unite[d]?\s*states?(\s*of\s*america)?$'
looks_like_us = (
    df_salary_all['country']
    .astype('string')
    .str.strip()
    .str.lower()
    .str.match(us_pattern, na=False)
)
df_salary_all.loc[looks_like_us, 'country'] = 'US'
df_salary_all = df_salary_all[df_salary_all['country'] == 'US']

df_salary_all['salary'] = pd.to_numeric(
    df_salary_all['salary'].astype(str).str.replace(',', '', regex=False),
    errors='coerce',
)

salary_by_state_all = (
    df_salary_all.groupby('state')['salary']
    .agg(['min', 'mean', 'max'])
    .sort_values('mean', ascending=False)
)
salary_by_state_all
```

### Bonus #2

Filter out the non-single-states (e.g., `'California, Colorado'`) in the most elegant way possible (i.e., *not* by blacklisting all the bad values).

```python
# Bonus #2: Filter out multi-state entries (e.g., 'California, Colorado') elegantly.

df_salary_all_single_state = df_salary_all[~df_salary_all['state'].str.contains(',', na=False)]
df_salary_all_single_state['state'].unique()
```

### Bonus #3

Show the quantiles instead of just minimum, mean, and maximum―say 0%, 5%, 25%, 50%, 75%, 95%, and 100%. Outliers may be deceiving.

Sort by whatever interests you―like say the *50th* percentile.

And throw in a count by state too. It would be interesting to know how many data points contribute to the figures for each state. (*Hint: your nice formatting from Bonus #1 might not work this time around.* 😜)

```python
# Bonus #3: Show quantiles (plus a count) by state instead of just min/mean/max.

quantiles = [0, 0.05, 0.25, 0.5, 0.75, 0.95, 1.0]

salary_quantiles_by_state = (
    df_salary_all_single_state
    .groupby('state')['salary']
    .quantile(quantiles)
    .unstack()
)
salary_quantiles_by_state.columns = [f'{int(q * 100)}%' for q in quantiles]

salary_quantiles_by_state['count'] = df_salary_all_single_state.groupby('state')['salary'].count()

salary_quantiles_by_state.sort_values('50%', ascending=False)
```
