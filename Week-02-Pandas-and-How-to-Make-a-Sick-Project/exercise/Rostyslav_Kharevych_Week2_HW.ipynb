### Clean the Data

Perhaps we’d like to know more another day, but the team is really interested in just the relationship between calories (`calories_day`) and weight. …and maybe gender.

Can you remove the other columns? (Assign the result to a new variable, `df_food_col_subset`.)

```python
# Remove ‘em.
df_food_col_subset = df_food[['calories_day', 'Gender', 'weight']].copy()
df_food_col_subset.shape
```

```python
# Is there a second way to remove columns?
df_food_col_subset = df_food.loc[:,['calories_day', 'Gender', 'weight']]
df_food_col_subset.shape
```

```python
# 🚀 Extra credit: What about a third way to remove columns?
keep_cols = ['calories_day', 'Gender', 'weight']
df_food_col_subset = df_food.drop(columns = [col for col in df_food.columns if col not in keep_cols])
df_food_col_subset.shape
```

What about `NaN`s? How many are there?

```python
# Count ‘em.
df_food_col_subset.isna().sum()
```

```python
# 🚀 Extra credit: Try to display the NaN sums as a nice pandas DataFrame.
df_food_col_subset.isna().sum().to_frame(name = 'Missing Cols')
```

```python
# 🚀 Extra credit: You can turn one line of code into multiple lines of code using a backslash as a continuation character to break up what would otherwise be a long line.
df_food_col_subset.isna().sum() \
    .reset_index() \
    .rename(columns={"index": "Column", 0:"Missing Cols"})
```

We gotta remove those `NaN`s―the entire row.

```python
# Drop ‘em in-place.
df_food_col_subset.dropna(inplace = True)
```

```python
# 🚀 Extra credit: Check if the in-place modifications also affected the original DataFrame, df_food. 🙂
df_food.isna().sum()
```

But what about the weird non-numeric values in the column obviously meant for numeric data?

Notice the data type of that column from when you got the types of all the columns?

If only we could convert the column to a numeric type and drop the rows with invalid values. 🤔

```python
# Fix that in-place.
df_food_col_subset['weight'] = pd.to_numeric(df_food_col_subset['weight'], errors = 'coerce')
df_food_col_subset.dropna(subset = "weight", inplace=True)

df_food_col_subset['weight'] = df_food_col_subset['weight'].astype(int)
```

```python
#Create a copy without filtering
df_usd = df_salary[df_salary['currency'] == "USD"].copy()

#Standardize US country variations
top_5_countries = df_usd['country'].value_counts().head(5).index.tolist()
df_usd['country'].replace(top_5_countries, 'US', inplace=True)

#Clean and convert salary to numeric
df_usd['salary'] = pd.to_numeric(
    df_usd['salary'].astype(str).str.replace('$', '', regex = False)\
    .str.replace(',', '', regex = False).str.strip(), errors = 'coerce')

#Filter out US and calculate the aggregation by state
df_usd[df_usd['country'] == "US"] \
    .groupby('state')['salary'] \
    .agg(["min", "mean", "max"]) \
    .sort_values(by = "mean", ascending = False)
```

```python
#Create a summary from a copy so that the original doesn't get touched
summary_table = (
    df_salary[df_salary['currency'] == 'USD'].copy()
    .assign(
        country=lambda d: d['country'].replace(d['country'].value_counts().head(5).index, 'US'),
        salary=lambda d: pd.to_numeric(d['salary'].astype(str).str.replace(r'[\$,]', '', regex=True), errors='coerce')
    )
    .query("country == 'US'")
    .groupby('state')['salary']
    .agg(['min', 'mean', 'max'])
    .sort_values(by='mean', ascending=False)
)

#Format using Styler
summary_table.style.format("${:,.2f}")
```
